**mutable vs immutable**
- mutables: characters in a string can be change by index operation
- immutable : characters in a string cannot be change index operation

```python
str1='welcome' #'l' to'L
str1[2]='L'
```

```python
str1='welcome' 
l1=['w','e','l','c','o','m','e']
l1[2]='L'
l1
```
