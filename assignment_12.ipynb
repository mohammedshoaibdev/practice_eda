```python
try:
    def sum(a,b,c):
        return a+b+c  
except Exception as e:
    print('error occured')
print(sum(5,6,77))
```

```python
try:
    def area_of_circle(pi,r):
        return pi*r*r
except Exception as e:
    print("enter correct value")
print(area_of_circle(3.14159,7))
```

```python
try:
    def number(num):
        if num%2==0:
            print('the number is even')
        else:
            print('the number is odd')
except Exception as e:
    print('enter the correct value')
number(53)
```

```python
try:
    def bmi(weight,hieght):
        bmi_calc=weight/ (hieght**2)
        return bmi_calc
except EXception as e:
    print("error occured please try again")
print(bmi(80,170))
```

```python
try:
    def greet(name,age):
        print(f'hello{name} you are {age} years old')
except Exception as e:
    print("an error occured")
greet('shoaib',30)
```

```python
try:
    def greater(a,b):
        if a>b:
            print(a)
        elif a<b:
            print(b)
        else:
            print("numbers should not be same")
except Exception as e:
    print("an error occured")
greater(9,90)
```

```python
try:
    def power(a,b):
        result=a**b
        return result 
except Exception as e:
    print("an error occured")
print(power(2,3))
```

```python
try:
    def simple_intrest(p,t,r):
        result=(p*r*t)/10
        return result
except Exception as e:
    print("an error occured")
print(simple_intrest(100,5,2))
```

```python
try:
    def leap_year(year):
        if year%4==0:
            print("it is leap year")
        else:
            print("it is not a leap year")
except Exception as e:
    print("an error occured")
leap_year(300)
```
