```python
try:
    def calculator(a,b):
        Sum=a+b
        sub=a-b
        mul=a*b
        div=a/b
        print(f'the addition of two number is {Sum} and the subtraction is {sub} and multiplication is {mul} and division is {div}',end=' ')
except Exception as e:
    print("an error occured")
calculator(15,6)
```

```python
try:
    def convert(weight_in_kgs):
        weight_in_pounds=weight_in_kgs*2.2
        final_weight=round(weight_in_pounds)
        print(f' the weight in pounds is {final_weight}')
except Exception as e:
    print("error occured")
convert(60)
```
