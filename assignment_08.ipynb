```python
try:
    n1=int(input("Enter a number1"))
    n2=int(input("Enter a number2"))
    
    def calc():
        sub=n1-n2
        mul=n1*n2
        div=n1/n2
        print(f" the subtraction is{sub} the multiplication is {mul}the division is {div}")
except Exception as e:
    print("error occured")
calc()
        
```

```python
try:
    pi=float(input("Enter pi value"))
    r=int(input("Enter radius"))
    def area_of_circle():
        result=pi*r*r
        print(f'the area of thr circle is {result}')
except Exception as e:
    print("enter correct value")
area_of_circle()
```

```python
try:
    fixed_value=.5
    breadth=float(input("Enter the breadth of triangle"))
    hieght=float(input("Enter the hieght of triangle"))
    def area_of_traingle():
        result=fixed_value*breadth*hieght
        print(result)
except Exception as e:
    print("error occured")
area_of_traingle()
```

```python
try:
    weight_in_kgs=eval(input("enter your weight"))
    def convert():
        weight_in_pounds=weight_in_kgs*2.2
        final_weight=round(weight_in_pounds)
        print(f' the weight in pounds is {final_weight}')
except Exception as e:
    print("error occured")
convert()
```
